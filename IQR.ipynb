```python
import pandas as pd
```

```python
df = pd.read_csv("trends.csv")
```

```python
df
```

```python
columnname='rank'
```

```python
Q1 = df[columnname].quantile(0.25)
```

```python
Q3 = df[columnname].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[(df[columnname] < lower) | (df[columnname] > upper)]
```

```python
clean_iqr = df[(df[columnname] >= lower) & (df[columnname] <= upper)]
```

```python
print(f"Cleaned DataFrame rows: {len(clean_iqr)}")
```

```python
print("Q1:", Q1)
```

```python
print("Q3:", Q3)
```

```python
print("IQR:", IQR)
```

```python
print("Lower Limit:", lower)
```

```python
print("Upper Limit:", upper)
```

```python
print("Outliers:")
print(outliers)
```

```python
print(outliers[columnname])
```

```python
print("Number of outliers:", len(outliers))
```

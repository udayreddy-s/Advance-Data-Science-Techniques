```python
# Remove duplicate rows
df = df.drop_duplicates().copy()

# Remove rows with missing target values
df = df.dropna(subset=["ATTORNEY"]).copy()

# Remove the identifier column
X = df.drop(columns=["ATTORNEY", "CASENUM"])
y = df["ATTORNEY"].astype(int)

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

display(X.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

```

```python

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("Training Percentage:", len(X_train) / len(X) * 100)
print("Testing Percentage:", len(X_test) / len(X) * 100)
```

```python
outlier_columns = ["CLMAGE", "LOSS"]
outlier_bounds = {}

for col in outlier_columns:
    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_bounds[col] = (lower_bound, upper_bound)

    outlier_count = (
        (X_train[col] < lower_bound) |
        (X_train[col] > upper_bound)
    ).sum()

    print(f"{col} outliers: {outlier_count}")
    print(f"Lower Bound: {lower_bound:.3f}")
    print(f"Upper Bound: {upper_bound:.3f}\n")

# Cap outliers using training-set bounds
for col, (lower, upper) in outlier_bounds.items():
    X_train[col] = X_train[col].clip(lower, upper)
    X_test[col] = X_test[col].clip(lower, upper)
```

```python
# Fill missing values using training data only
imputer = SimpleImputer(strategy="median")

X_train_imputed = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test_imputed = pd.DataFrame(
    imputer.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

# Scale the features
scaler = StandardScaler()

X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train_imputed),
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test_imputed),
    columns=X_test.columns,
    index=X_test.index
)

print("Missing values after imputation:")
print(X_train_scaled.isnull().sum())

print("\nTraining data after preprocessing:")
display(X_train_scaled.head())
```

```python
print("Selected Features:")
print(X_train_scaled.columns.tolist())

print("\nTarget Variable:", y.name)
```

```python
# Create the model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
coefficients = pd.DataFrame({
    "Feature": X_train_scaled.columns,
    "Coefficient": model.coef_[0]
})

print("Model Coefficients:")
display(coefficients)

print("Model Intercept:", model.intercept_[0])
```

```python
# Predict target values
y_pred = model.predict(X_test_scaled)

print("First 10 Actual Values:")
print(y_test.iloc[:10].tolist())

print("\nFirst 10 Predicted Values:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy:", round(accuracy * 100, 2), "%")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Attorney (0)", "Attorney (1)"],
    yticklabels=["No Attorney (0)", "Attorney (1)"]
)

plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

comparison["Correct"] = comparison["Actual"] == comparison["Predicted"]

display(comparison.head(20))
```

```python
print("Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

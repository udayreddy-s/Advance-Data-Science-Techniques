```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("Mall_Customers.csv")
```

```python
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df = df.dropna()
```

```python
df["Genre"] = df["Genre"].astype(str).str.strip().str.title()
```

```python
df.to_csv("Cleaned_Mall_Customers.csv", index=False)
```

```python
df["Annual Income"] = df["Annual Income (k$)"] * 1000
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["Under 18", "18-30", "31-45", "46-60", "60+"]
)
```

```python
scaler = MinMaxScaler()
```

```python
columns = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
```

```python
for col in columns:
    df[col + " Normalized"] = scaler.fit_transform(df[[col]])
```

```python
print("Missing values:")
```

```python
print(df.isnull().sum())
```

```python
print("\nDuplicates:", df.duplicated().sum())
```

```python
print("\nGenre values:", df["Genre"].unique())
```

```python
normalized = [
    "Age Normalized",
    "Annual Income (k$) Normalized",
    "Spending Score (1-100) Normalized"
]
```

```python
print("\nNormalized range:")
```

```python
print(df[normalized].agg(["min", "max"]))
```

```python
df.to_csv("Transformed_Mall_Customers.csv", index=False)
```
